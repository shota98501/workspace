```python
import pandas as pd
```

```python
df = pd.read_csv('month_total_sales.csv')
df
```

```python
df['month'] = pd.to_datetime(df['month'])
```

```python
df['year'] = df['month'].dt.year
df['month_num'] = df['month'].dt.month
```

```python
X = df[['year', 'month_num']]
y = df['total_sales']
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

model.fit(X,y)
```

```python
future = pd.DataFrame({
    "year": [2025, 2025, 2025],
    "month_num": [1, 2, 3]
})
```

```python
future["predicted_sales"] = model.predict(future)
future
```

```python
future.to_csv(
    "sales_forecast.csv",
    index=False
)
```
