```python
import pandas as pd
```

```python
df = pd.read_csv('vgsales.csv')
```

```python
df
```

```python
df.isnull().sum()
```

```python
year = df.groupby("Platform")["Year"].median()

def fill_year(row):
    if pd.isna(row["Year"]):
        plat = row["Platform"]
        if plat in year and not pd.isna(year[plat]):
            return int(year[plat])
        return row["Year"]
    return row["Year"]

df["Year"] =df.apply(fill_year, axis=1) 
```

```python
df.isna().sum()
```

```python
name_pub_mode = df.groupby("Name")["Publisher"].agg(
    lambda x: x.mode().iloc[0] if x.mode().size > 0 else None
)

def fill_publisher(row):
    if pd.isna(row["Publisher"]):
        name = row["Name"]
        if name in name_pub_mode and pd.notna(name_pub_mode[name]):
            return name_pub_mode[name]
        return "Unknown"
    return row["Publisher"]

df["Publisher"] = df.apply(fill_publisher, axis=1)
```

```python
df.isna().sum()
```

```python
df =df.drop_duplicates()
```
