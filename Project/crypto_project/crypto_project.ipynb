```python
import time
import requests
import pandas as pd
```

```python
coin_ids = ["bitcoin","ethereum","solana"]
url = "https://api.coingecko.com/api/v3/simple/price"


all_coin_data = []

print("Starting CoinGecko API data collection...")

for coin in coin_ids:
    print(f"Fetching data for {coin}...")

    params = {
    "ids": coin,
    "vs_currencies": "jpy",
    }

    try:
        response = requests.get(url, params=params)

        if response.status_code == 200:
            data = response.json()

            if coin in data:
                coin_info = data[coin]

                row = {
                    "Coin" : coin,
                    "Price": coin_info.get("jpy"),

                }
                all_coin_data.append(row)
        else:
            print(
                f"Failed to fetch {coin}. HTTP Status Code: {response.status_code}"
            )
    except Exception as e:
       print(f"An error occurred while fetching {coin}: {e}")

    time.sleep(2.0)

df = pd.DataFrame(all_coin_data)

print("\nData collection complete!")
print(df)
```
