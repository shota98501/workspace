```python
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
df
```

duplicates are removed

```python
df.drop_duplicates(inplace=True)
df
```

"eight" becomes 8 

```python
df["distance_km"] = df["distance_km"].replace("eight",8).pipe(pd.to_numeric, errors="coerce")
df
```

numeric columns are converted correctly

```python
numeric_columns = [
    "distance_km",
    "item_count",
    "delivery_minutes"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )
```

negative distance and item counts become missing

```python
df[["distance_km","item_count"]] = df[["distance_km","item_count"]].mask(df[["distance_km","item_count"]] < 0)
df
```

peak_hour becomes Boolean

```python
df["peak_hour"] = (df["peak_hour"].str.strip().str.lower().map({"yes":True,"no":False}))
df
```

weather values have consistent capitalization & "unknown" weather becomes Unknown

```python
df["weather"] = df["weather"].str.capitalize()
df
```

data_issue flags missing distance, item count, or delivery time

```python
required_columns = [
    "distance_km",
    "item_count"
]

df["data_issue"] = df[required_columns].isna().any(axis=1)
df
```

```python
df =df.fillna(0)
df
```

```python
df.to_csv("cleaned_data.csv")
```
