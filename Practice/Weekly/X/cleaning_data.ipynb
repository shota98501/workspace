```python
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
df
```

duplicates are removed

```python
df.drop_duplicates(inplace=True)
df
```

"twelve" becomes 12

```python
df["distance_km"] = df["distance_km"].replace("twelve",12)
df
```

numeric columns are converted correctly

```python
numeric_columns = [
    "distance_km",
    "duration_minutes",
]

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

df
```

negative distances become missing

```python
df[["distance_km","duration_minutes"]] = df[["distance_km","duration_minutes"]].mask(df[["distance_km","duration_minutes"]] < 0)
df
```

```python
df[["distance_km","duration_minutes"]] = df[["distance_km","duration_minutes"]].fillna(0)
df
```

rider types become Member, Casual, or Unknown

```python
df["rider_type"] = df["rider_type"].str.strip().str.capitalize().fillna("Unknown")
df
```

weekend becomes Boolean

```python
df["weekend"] = df["weekend"].str.strip().str.lower().map({"yes":True,"no":False})
df
```

weather values use consistent capitalization

```python
df["weather"] = df["weather"].str.strip().str.capitalize()
df
```

```python
df.to_csv("cleaned_data.csv")
```
