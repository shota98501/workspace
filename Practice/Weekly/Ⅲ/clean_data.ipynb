```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data.csv')
df
```

```python
df.info
```

```python
df.drop_duplicates(inplace=True)
df
```

```python
df["join_date"] = df["join_date"].apply(lambda x:None if x=="invalid" else pd.to_datetime(x, format="mixed"))
df
```

```python
df["status"] = df["status"].str.strip().str.capitalize()
df
```

```python
df["age"] = df["age"].replace({"thirty":30, "-2":None}).pipe(pd.to_numeric, errors="coerce")
df
```

```python
df["monthly_fee"] = df["monthly_fee"].str.replace(",","").pipe(pd.to_numeric, errors="coerce")
df
```

```python
df2 = pd.read_csv("data2.csv")
df2
```

```python
all_df = pd.merge(df,df2.drop(columns=["monthly_fee"]), on="member_id", how="inner")
all_df
```

```python
all_df.info()
```

```python
all_df["visit_date"] = pd.to_datetime(all_df["visit_date"], format="mixed")
all_df
```

```python
df = all_df[["member_id","visit_date","join_date","age","plan_type","status","monthly_fee"]]
```
