```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data.csv')
df
```

```python
df.drop_duplicates(inplace=True)
```

```python
df["delivery_date"] = df["delivery_date"].replace("invalid",None).pipe(pd.to_datetime, format="mixed")
df
```

```python
df["supplier"] = df["supplier"].str.strip().str.capitalize()
df
```

```python
df["units_received"] = df["units_received"].replace({"ten":10,"-":""},regex=True).pipe(pd.to_numeric)
df
```

```python
df["unit_cost"] = df["unit_cost"].replace({"unknown":None,",":""},regex=True).apply(pd.to_numeric,errors = 'coerce')
df
```

```python
df["data_issue"] = df[
    ["supplier", "delivery_date", "unit_cost"]
].isna().any(axis=1)
df
```

```python
df2 = pd.read_csv("data2.csv")
df2
```

```python
df3 = pd.read_csv("data3.csv")
df3
```

```python
all_df = pd.merge(pd.merge(df,df2,on="product_id",how="outer"),df3,on="product_id", how="outer")
all_df
```
