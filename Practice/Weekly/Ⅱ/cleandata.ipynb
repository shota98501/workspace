```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data.csv')
df
```

```python
df = df.drop_duplicates()
```

```python
df
```

```python
df["courier"] = df["courier"].str.capitalize()
df["courier"] = df["courier"].fillna("Unknown")
df
```

```python
df["status"] = df["status"].str.replace("delivred", "Delivered")
df["status"] = df["status"].str.capitalize()
df
```

```python
df["delivery_minutes"] = df["delivery_minutes"].replace({"-5":"5", "eighty":"80"})
df
```

```python
df["delivery_minutes"] = pd.to_numeric(df["delivery_minutes"])
df
```

```python
df["delivery_minutes"] = df["delivery_minutes"].fillna(0)
```

```python
df.to_csv('cleaneddata.csv')
```

```python
df1 = pd.read_csv('postgredata.csv')
df1
```

```python
total_df = pd.merge(df, df1, on="delivery_id", how="inner")
total_df
```

```python
total_df.info()
```

```python
total_df = total_df.astype({"promised_date":"datetime64[ns]","delivered_date":"datetime64[ns]" })
total_df
```
