```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data.csv')
df
```

```python
df["courier"] = df["courier"].str.capitalize()
df["courier"] = df["courier"].fillna("Unknown")
df
```

```python
df["status"] = df["status"].str.replace("delivred", "Delivered")
df["status"] = df["status"].str.capitalize()
df
```

```python
df["delivery_minutes"] = df["delivery_minutes"].replace({"-5":"5", "eighty":"80"})
df
```

```python
df["delivery_minutes"] = pd.to_numeric(df["delivery_minutes"])
df
```
