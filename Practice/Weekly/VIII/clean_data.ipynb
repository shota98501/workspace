```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data.csv')
df
```

```python
df.drop_duplicates(inplace=True)
df
```

```python
df["tenure_months"] = df["tenure_months"].replace({"twelve":12,"-":""},regex=True).pipe(pd.to_numeric)
df
```

```python
df["monthly_charge"] = df["monthly_charge"].replace({",":"","unknown":None},regex=True).pipe(pd.to_numeric)
df
```

```python
df["contract"] = df["contract"].str.strip().str.capitalize()
df
```
