```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data.csv')
df
```

```python
df["return_date"] = df["return_date"].replace("invalid",None).pipe(pd.to_datetime, format="mixed")
df
```

```python
df["reason"] = df["reason"].str.strip().str.capitalize()
df
```

```python
df["refund_amount"] = df["refund_amount"].replace({",": "", "unknown": None}, regex=True).pipe(pd.to_numeric, errors='coerce')
df
```

```python
df["approved"] = np.where(df["approved"].str.lower().str.startswith("y", na=False), "Yes", "No")
df
```
