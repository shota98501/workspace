```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data.csv')
df
```

```python
df.drop_duplicates(inplace=True)
df
```

```python
df['start_date'] = df['start_date'].replace('invalid',None).pipe(pd.to_datetime,format='mixed')
df
```

```python
df['ad_spend'] = df['ad_spend'].replace({',':'','unknown':None}, regex=True).pipe(pd.to_numeric, errors = 'coerce')
df
```

```python
df['impressions'] = df['impressions'].replace({",":"","-":""}, regex=True).pipe(pd.to_numeric, errors='coerce')
df
```

```python
df['status'] = df['status'].str.strip().str.capitalize()
df
```
