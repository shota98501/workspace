```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Data.csv")
df
```

removes duplicate rows

```python
df.drop_duplicates()
```

converts visit_date to datetime

```python
df["visit_date"] = df["visit_date"].apply(lambda x:"NaT" if x=="invalid" else pd.to_datetime(x, format="mixed"))
df
```

converts visit_date to datetime, using NaT for invalid dates

```python
df["rating"] = df["rating"].fillna(df["rating"].mean())
df
```

converts amount to numeric

```python
df["amount"] = (df["amount"].astype(str).str.replace(",", "").str.replace('"', ""))
df["amount"] = pd.to_numeric(df["amount"], errors="coerce")
df["amount"] = df["amount"].astype("Int64")
df
```

standardizes member to Yes or No

```python
df["member"] = np.where(df["member"].str.lower().str.startswith("y", na=False), "Yes", "No")
df
```
