```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data.csv')
df
```

```python
df.drop_duplicates(inplace=True)
df
```

```python
df["satisfaction_score"] = df["satisfaction_score"].replace({"6":None, "two":2}).pipe(pd.to_numeric, errors="coerce")
df
```

```python
df[["contact_channel","issue_type"]] = df[["contact_channel","issue_type"]].apply(lambda x: x.str.strip().str.capitalize())
df
```

```python
df2 = pd.read_csv('data2.csv')
df2
```

```python
all_df = pd.merge(df,df2, on ="ticket_id", how="inner")
all_df
```

```python
all_df.info()
```

```python
all_df[["opened_at","closed_at"]] = all_df[["opened_at","closed_at"]].apply(pd.to_datetime, format="mixed")
all_df
```

```python
all_df.info()
```

```python
df = all_df[["ticket_id","agent_id","satisfaction_score","contact_channel","issue_type","opened_at","closed_at","priority"]]
df
```

```python
df.to_csv("complete_data.csv")
```
